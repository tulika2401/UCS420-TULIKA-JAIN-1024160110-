```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print(reversed_arr)

def find_most_frequent_and_indices(a):
    counts = np.bincount(a)
    most_freq_val = np.argmax(counts)
    indices = np.where(a == most_freq_val)[0]
    return most_freq_val, indices

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
freq_val_x, indices_x = find_most_frequent_and_indices(x)
print(freq_val_x, indices_x)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
freq_val_y, indices_y = find_most_frequent_and_indices(y)
print(freq_val_y, indices_y)
```
